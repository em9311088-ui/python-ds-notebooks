# 03 — Collections and Comprehensions

## Learning goals
- Work with lists, tuples, dictionaries, and sets
- Use list/dict comprehensions

```python
fruits = ["apple", "banana", "mango", "apple"]
fruit_set = set(fruits)
print("KABABA FRUITS:", fruit_set)

prices = {"apple": 1.2, "banana": 0.8, "mango": 1.5}
print("Mango price:", prices["mango"])
```

```python
numbers = list(range(1, 11))
squares = [n**2 for n in numbers]
even_squares = [n**2 for n in numbers if n % 2 == 0]

print("Squares:", squares)
print("Even squares:", even_squares)
```

```python
# Build dictionary with keys 1–5 and their cubes as values
cubes = {}

for i in range(1, 6):
    cubes[i] = i ** 3

print(cubes)
```
